```python
#Remove Outliers using IQR Method
Q1 = df["Avg. Area Income"].quantile(0.25)
Q3 = df["Avg. Area Income"].quantile(0.75)

IQR = Q3 - Q1

lower = Q1 - 1.5*IQR
upper = Q3 + 1.5*IQR

df = df[(df["Avg. Area Income"] >= lower) & (df["Avg. Area Income"] <= upper)]
```

```python
sns.boxplot(x=df["Avg. Area Income"])
plt.show()
```

```python
#Remove Outliers using IQR Method
Q1 = df["Price"].quantile(0.25)
Q3 = df["Price"].quantile(0.75)

IQR = Q3 - Q1

lower = Q1 - 1.5*IQR
upper = Q3 + 1.5*IQR

df = df[(df["Price"] >= lower) & (df["Price"] <= upper)]
```

```python
sns.boxplot(x=df["Price"])
plt.show()
```

## SLR

```python
X = df[['Avg. Area Income']]
y = df['Price']
```

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(X,y,test_size=0.20, random_state = 42)
```

```python
from sklearn.linear_model import LinearRegression

slr_model = LinearRegression()

slr_model.fit(X_train, y_train)
```

```python
y_pred = slr_model.predict(X_test)
```

```python
from sklearn.metrics import mean_absolute_error
from sklearn.metrics import mean_squared_error
from sklearn.metrics import r2_score
from sklearn.metrics import root_mean_squared_error
import numpy as np

print("Simple Linear Regression Results")
print("------------------------------")

print("MAE :", mean_absolute_error(y_test,y_pred))
print("MSE :", mean_squared_error(y_test,y_pred))
print("RMSE :", root_mean_squared_error(y_test,y_pred))
print("R^2 :", r2_score(y_test,y_pred))
```

```python
print("Intercept: ",slr_model.intercept_)
print("Slope: ", slr_model.coef_[0])
```

```python
plt.figure(figsize=(8,6))
plt.scatter(X_test, y_test, color='blue', label='Actual Data')
plt.plot(X_test, y_pred, color='red', linewidth=2, label = 'Regression Line')

plt.xlabel("Average Area Income")
plt.ylabel("House Price")
plt.title("Simple Linear Regression")
plt.legend()

plt.show()
```

```python
income = float(input("Enter Average Area Income: "))
new_data = np.array([[income]])
prediction = slr_model.predict(new_data)
print("Predicted House Price = ${:.2f}".format(prediction[0]))
```

# MLR

```python
X = df[['Avg. Area Income', 'Avg. Area House Age', 'Avg. Area Number of Rooms', 
        'Avg. Area Number of Bedrooms', 'Area Population']]
y = df['Price']
```

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(X,y,test_size=0.20, random_state = 42)
```

```python
from sklearn.linear_model import LinearRegression

mlr_model = LinearRegression()

mlr_model.fit(X_train, y_train)
```

```python
coef = pd.DataFrame(
    mlr_model.coef_,
    X.columns,
    columns = ["Coeeficient"]
)

coef
```

```python
y_pred = mlr_model.predict(X_test)
```

```python
print("Multiple Linear Regression Results")
print("------------------------------")

print("MAE :", mean_absolute_error(y_test,y_pred))
print("MSE :", mean_squared_error(y_test,y_pred))
print("RMSE :", root_mean_squared_error(y_test,y_pred))
print("R^2 :", r2_score(y_test,y_pred))
```

```python
print("Intercept: ",mlr_model.intercept_)
print("Slope: ", mlr_model.coef_[0])
```

```python
plt.figure(figsize=(6,6))
plt.scatter(y_test, y_pred)
plt.plot([min(y_test), max(y_test)], [min(y_test), max(y_test)] , color = 'red', linewidth = 2, label = 'Perfect Fit')
plt.xlabel("Actual Price")
plt.ylabel("Predicted Price")
plt.title("Actual vs Predicted")

plt.show()
```

```python
income = float(input("Enter Average Area Income: "))
house_age = float(input("Enter Average House Age: "))
no_of_rooms = float(input("Enter Average Area Number of rooms: "))
no_of_bedrooms = float(input("Enter Average Area Number of bedrooms: "))
area_pop = float(input("Enter Area Population: "))
test_input = np.array([[income, house_age, no_of_rooms, no_of_bedrooms, area_pop]])
predicted_price = mlr_model.predict(test_input)
print("Predicted House Price = ${:.2f}".format(predicted_price[0]))
```

## Ridge Regression

```python
from sklearn.model_selection import GridSearchCV
```

```python
from sklearn.linear_model import Ridge

ridge = Ridge()

ridge.fit(X_train, y_train)

ridge_pred = ridge.predict(X_test)

print("Default Ridge R2 :", r2_score(y_test, ridge_pred))
```

```python
param_grid = {'alpha' : [0.001, 0.01, 0.1, 1, 10, 100]}
```

```python
grid_ridge = GridSearchCV(estimator = Ridge(), param_grid = param_grid , scoring = 'r2', cv=5)
```

```python
grid_ridge.fit(X_train, y_train)
```

```python
print("Best Alpha :", grid_ridge.best_params_)
```

```python
print(grid_ridge.best_score_)
```

```python
best_ridge = grid_ridge.best_estimator_

ridge_pred = best_ridge.predict(X_test)
```

```python
print("MAE :", mean_absolute_error(y_test,ridge_pred))
print("MSE :", mean_squared_error(y_test,ridge_pred))
print("RMSE :", root_mean_squared_error(y_test,ridge_pred))
print("R^2 :", r2_score(y_test,ridge_pred))
```

```python
print(grid_lasso.best_params_)
```
